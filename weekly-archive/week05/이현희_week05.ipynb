1. 그래프 기본 모티프 사슬 / 분기 / 충돌부

## 1) 사슬(Chain) — `X → M → Y`
- **의미**: X가 **매개변수 M**을 거쳐 Y에 영향을 주는 연속 경로.
- **기본 상태**: `X ⟂̸ Y` (의존).
- **조건화 효과**: `M`에 조건화하면 **경로 차단** → `X ⟂ Y | M`.
- **실무 가이드**
  - **총효과** 추정: `M` **조정 금지**(매개효과까지 포함).
  - **직접효과** 추정: `M` **조정**(매개경로 제거).
- **예시**: `광고노출(X) → 인지도(M) → 구매(Y)`

---

## 2) 분기(Fork, 공통원인 = 교란 요인) — `U → X`, `U → Y`
- **의미**: `U`가 X·Y의 **공통 원인(혼란/교란변수)**.
- **왜 문제인가? (교란)**: `U` 때문에 X–Y에 **겉보기 상관**이 생겨, X→Y 효과 추정이 **편향**됨.
- **조건화 효과**: `U`를 조건화하면 **백도어 경로 차단** → `X ⟂ Y | U`.
- **조정 원칙**
  - 목표는 X→Y **모든 백도어 경로 차단**.
  - **최소 충분 조정집합**: 불필요한 변수 없이 **필요한 공통원인만** 조정.
  - `U`의 **자손(대리변수)** 만 조정하면 **잔여 교란**이 남을 수 있음 → 가능하면 `U` 그 자체를 조정.
- **예시**: `성수기(U) → 프로모션(X)`, `성수기(U) → 매출(Y)`

---

## 3) 충돌부(Collider) — `X → C ← Y`  (*설명경쟁 / explaining away*)
- **의미**: `C`는 X·Y의 **합류점**(두 원인이 만나 **결과를 함께 설명**).
- **기본 상태**: `X ⟂ Y` (독립).
- **조건화 효과**: `C` **또는 자손**을 조건화하면 **경로 개방** → `X ⟂̸ Y | C`  
  (버크슨/선택 편향; **다른 요인에 의해 설명되는 현상**이 발생: C가 고정되면 X가 낮을수록 Y가 높아져야 C를 설명함 → X–Y 역상관 유도 가능)
- **실무 가이드**
  - **콜라이더/자손 조정 금지**(선택·필터 변수 주의).
  - 분석 대상 선정 기준이 **콜라이더의 자손**이 아닌지 점검.
- **예시**
  - `실력(X) → 합격(C) ← 인맥(Y)` → 합격자만 보면 X–Y에 **거짓 상관**.
  - `질병중증(X) → 입원(C) ← 접근성(Y)` → 입원자만 보면 **왜곡**.

---

## 한눈 표

| 구조 | 형태 | 기본 상태 | 조건화 대상 | 조건화 후 | 실무 포인트 |
|---|---|---|---|---|---|
| Chain | `X → M → Y` | `X ⟂̸ Y` | **M** | `X ⟂ Y \\| M` (**차단**) | 총효과=**M 비조정**, 직접효과=**M 조정** |
| Fork(교란) | `U → X`, `U → Y` | `X ⟂̸ Y` | **U** | `X ⟂ Y \\| U` (**백도어 차단**) | **최소 충분 조정집합**으로 공통원인만 조정 |
| Collider(충돌부) | `X → C ← Y` | `X ⟂ Y` | **C(또는 자손)** | `X ⟂̸ Y \\| C` (**개방/설명경쟁**) | **콜라이더/자손 조정 금지**, 선택편향 주의 |



# 종속성 흐름(의존성) 치트 시트 (요약)

- **체인/분기**: 중간 노드 조건화 → **해당 경로 차단**.  
- **콜라이더(및 자손)**: 조건화 → **해당 경로 개방**(버크슨/설명경쟁).  
  - 총효과: **매개(M) 조정 금지**, 공통원인(U)만 조정.  
  - 직접효과: **매개(M) 조정** + 공통원인(U) 조정.  
  - 콜라이더/자손 조정 금지.



# 보정(Adjustment) 공식 — 백도어 표준화 & 가중치

## 1) 조건부 독립성(무혼란/교환가능성)
- 표기: \(Y(x) \perp\!\!\!\perp X \mid Z\) — \(Z\)를 고정하면 처치 X가 **as-if 무작위**.

## 2) (백도어) 표준화(Standardization)
- **이산 \(Z\)**  
  \[
  P\!\big(Y{=}y \mid do(X{=}x)\big) \;=\; \sum_{z} P\!\big(Y{=}y \mid X{=}x, Z{=}z\big)\, P(Z{=}z)
  \]


- **ATE**
  \[
  \text{ATE} = E[Y \mid do(1)] - E[Y \mid do(0)]
  \]

## 3) 성향점수 가중치(IPW) — 동일 타깃 분포 재구성
- ** 처치 \(x\)**  
  \[
  \widehat{\tau}(x) = E\!\left[\frac{\mathbf{1}\{X{=}x\} \, Y}{\hat e_x(Z)}\right], \quad \hat e_x(Z) = \widehat{P}(X{=}x \mid Z)
  \]




# 양수성(Overlap/Positivity) 가정

## 정의
- (이산 처치) 모든 \(Z=z\)에 대해 \(0 < P(X=1 \mid Z=z) < 1\).  

## 필요성
- 같은 \(Z\)에서 **처치/비처치가 모두 가능**해야 **비교 가능한 대조군**이 존재.  
- 위반 시 **외삽 의존**, **가중치 폭주**, 추정 **불안정**.

## 유의사항
1) **공통지지 제한/트리밍**: \(\alpha \le \hat e(Z) \le 1-\alpha\).  
2) **안정화/Overlap 가중치**(SIPW, overlap weighting).  
3) **매칭/칼리퍼/CEM**로 겹치지 않는 관측치 제거.  
4) **타깃 전환**: ATE → ATT/ATC/ATO.  
5_ **설계 단계**: 무작위화/블로킹으로 양수성 확보.



# 편향(Bias) — 1)교란편향 2) 선택편향

## 1) 교란편향(Confounding Bias)
- **핵심**: **열린 뒷문경로(backdoor path)** 에서 발생 — **처치 \(X\)** 와 **결과 \(Y\)** 가 **같은 원인 \(U\)** 를 공유하기 때문.
  - DAG: `U → X`, `U → Y` (Fork). 이때 `X ← U → Y` 경로가 열려 있으면 편향.
- **그래프 규칙**: X→Y **모든 백도어 경로 차단**(체인/분기 조건화=차단, 콜라이더/자손 조건화=개방 주의).
- **보정 방법**: 표준화, 성향점수(IPW/매칭), DR, 사전 층화/블로킹.
- **실무 A — 대리교란요인(Proxy Confounders)**: \(U\) 미관측 시, \(U\)와 연결된 대리 \(W\)로 부분 보정(잔여 교란 가능 → 민감도 분석).
- **실무 B — 랜덤화 재해석(Randomization Reinterpretation)**: \(Z\)를 충분히 세분해 **층 내 as-if 무작위**; 층화 표준화/매칭으로 구현. **양수성** 확보 필수.

## 2) 선택편향(Selection Bias)
- **정의**: **선택/필터 변수 \(S\)** 에 **조건화**하여 **콜라이더 경로**가 **개방**되며 발생(버크슨/설명경쟁).
  - 예: `X → S ← Y`에서 \(S\) 또는 자손에 조건화 ⇒ `X ⟂̸ Y \mid S`.
- **발생 패턴**: 합격자/구매자/입원자만 분석(표본선정), 사후변수 조건화, 누락/검열 메커니즘.
- **대응**: 설계(선택 기준 완화, RCT/블라인드), 분석(콜라이더/자손 조정 금지, 선택·검열 가중치/결측모형, 감도 분석).
